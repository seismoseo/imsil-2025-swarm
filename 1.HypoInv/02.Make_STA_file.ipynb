## Import libraries

```python
import numpy as np
import pandas as pd
from glob import glob
from obspy import read
import os
```

## Make used station catalog

```python
path_to_dir="/home/msseo/works/72.2025_Imsil_swarm/mainshock_waveforms/20*/"
SAC_FILE_LIST=glob(path_to_dir+"*sac")
used_station_list=[]
for sacfile in SAC_FILE_LIST:
    staname=sacfile.split("/")[-1].split(".")[2]
    if staname not in used_station_list:
        used_station_list.append(staname)
    else: pass
print(f"Number of used stations: {len(used_station_list)}")
used_station_array=np.array(sorted(used_station_list))

# Make array of 1) station name, 2) station latitude, 3) station longitude, and 4) station elevation in same order
network_array=np.zeros(len(used_station_array)).astype("str")
stla_array=np.zeros(len(used_station_array))
stlo_array=np.zeros(len(used_station_array))
stel_array=np.zeros(len(used_station_array))
weight_array=np.ones(len(used_station_array))*100.
for sacfile in SAC_FILE_LIST:
    try:
        tr=read(sacfile)[0]
        staname=sacfile.split("/")[-1].split(".")[2]
        stla=tr.stats.sac.stla
        stlo=tr.stats.sac.stlo
        # stel=tr.stats.sac.stel
        # network=tr.stats.network
        stel, network=0.0, "KS"
        network_array[np.where(used_station_array==staname)[0][0]]=network
        stla_array[np.where(used_station_array==staname)[0][0]]=stla
        stlo_array[np.where(used_station_array==staname)[0][0]]=stlo
        stel_array[np.where(used_station_array==staname)[0][0]]=stel
        # print(f"{network} {staname}")
    except:
        pass
    
networkcode_array=[]
for i in range(len(used_station_array)):
    networkcode_array.append(network_array[i]+"."+used_station_array[i])
networkcode_array=np.array(networkcode_array)
    
# 3) Make DataFrame using arrays generated in previous step
used_station_catalog=pd.DataFrame(columns=["NetworkCode", "Latitude", "Longitude", "Elevation", "Weight"])
used_station_catalog["NetworkCode"]=networkcode_array
used_station_catalog["Latitude"]=stla_array
used_station_catalog["Longitude"]=stlo_array
used_station_catalog["Elevation"]=stel_array
used_station_catalog["Weight"]=weight_array

used_station_catalog.to_csv("/home/msseo/works/72.2025_Imsil_swarm/1.HypoInv/STA/Imsil.sta", header=False, index=False)
```
