## Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

## Load entire station table (KMA)

```python
kma_stations_100km=pd.read_csv("/home/msseo/works/72.2025_Imsil_swarm/station_table/used_stations_100km.csv", sep=",")
kma_stations_100km.head()
```

## Select stations to use for detection and save as `*.csv` file

```python
selected_stations=np.array(["CHOA", "GKSA", "IMSB", "JASA", "JEU2", "JLSA", "MALA", "NAWB", "SIDA", "SUCA", "TAIA"])
print(f"Number of stations: {len(selected_stations)}")
```

```python
detection_stations=pd.read_csv("./stations.csv", sep=",", names=["id", "network", "station", "channel", "latitude", "longitude", "unit", "component", "response"])
detection_stations
```

```python
id_list, network_list, station_list, channel_list=[], [], [], []
stla_list, stlo_list,  unit_list, comp_list, resp_list=[], [], [], [], []

for code in selected_stations:
    network=kma_stations_100km[kma_stations_100km["Code"]==code].reset_index(drop=True)["Network"][0]
    channel=kma_stations_100km[kma_stations_100km["Code"]==code].reset_index(drop=True)["Sensor"][0]
    stla=kma_stations_100km[kma_stations_100km["Code"]==code].reset_index(drop=True)["Latitude"][0]
    stlo=kma_stations_100km[kma_stations_100km["Code"]==code].reset_index(drop=True)["Longitude"][0]
    comp="E,N,Z"
    id_=f"{network}.{code}..{channel}"
    
    if channel=="HG":
        unit="m/s**2"
    else:
        unit="m/s"
       
    id_list.append(id_)
    network_list.append(network)
    station_list.append(code)
    channel_list.append(channel)
    stla_list.append(stla)
    stlo_list.append(stlo)
    unit_list.append(unit)
    comp_list.append(comp)
    
detection_stations["id"]=id_list
detection_stations["network"]=network_list
detection_stations["station"]=station_list
detection_stations["channel"]=channel_list
detection_stations["latitude"]=stla_list
detection_stations["longitude"]=stlo_list
detection_stations["unit"]=unit_list
detection_stations["component"]=comp_list
```
