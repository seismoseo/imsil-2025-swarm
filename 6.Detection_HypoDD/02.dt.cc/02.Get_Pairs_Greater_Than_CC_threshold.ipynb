## Import libraries

```python
import numpy as np
import pandas as pd
from glob import glob
from tqdm import tqdm
```

## Combine fragmented `dt.cc` files for each event pair

**`dt.cc_P`**

```python
# Combine each dt.cc_P files for individual event pair into single file
path_to_file="/home/msseo/works/72.2025_Imsil_swarm/6.Detection_HypoDD/02.dt.cc/dt.cc_P/"
file_list=glob(path_to_file+"dt.cc_P_*")
f=open(path_to_file+"dt.cc_P_combined", 'w')
for file in tqdm(file_list):
    g=open(file, 'r')
    lines=g.read()
    g.close()
    
    if len(lines.split("\n"))>2:
        f.write(lines)
    else: pass
    
f.close()
```

**`dt.cc_S`**

```python
# Combine each dt.cc_S files for individual event pair into single file
path_to_file="/home/msseo/works/72.2025_Imsil_swarm/6.Detection_HypoDD/02.dt.cc/dt.cc_S/"
file_list=glob(path_to_file+"dt.cc_S_*")
f=open(path_to_file+"dt.cc_S_combined", 'w')
for file in tqdm(file_list):
    g=open(file, 'r')
    lines=g.read()
    g.close()
    
    if len(lines.split("\n"))>2:
        f.write(lines)
    else: pass
    
f.close()
```

## Get pairs such that $CC\geq0.7$

**`dt.cc_P`**

```python
# Extract only pairs with CC>=0.7
threshold=0.7
path_to_file="/home/msseo/works/72.2025_Imsil_swarm/6.Detection_HypoDD/02.dt.cc/dt.cc_P/"
f1=open(path_to_file+f"dt.cc_P_{threshold}", 'w')
f2=open(path_to_file+"dt.cc_P_combined", 'r')
lines=f2.readlines()
for i in tqdm(range(len(lines))):
    if lines[i][0]=="#":
        f1.write(lines[i])
    else:
        try:
            if float(lines[i][23:34].replace(" ", ""))>=threshold and float(lines[i][23:34].replace(" ", ""))<=1:
                f1.write(lines[i])
            else: pass
        except: pass
f2.close()
f1.close()
```

```python
# Clean up dt.cc_P_0.7
f1=open(path_to_file+f"dt.cc_P_{threshold}", 'r')
f2=open("/home/msseo/works/72.2025_Imsil_swarm/6.Detection_HypoDD/02.dt.cc/"+f"dt.cc_P_{threshold}", 'w')
lines=f1.readlines()
for i in range(len(lines)-1):
    if lines[i][0]=="#" and lines[i+1][0]=="#":
        pass
    else:
        f2.write(lines[i])
f2.close()
f1.close()
```
