## Import libraries

```python
from glob import glob
import os, shutil
from tqdm import tqdm
```

## Change the SAC file name

```python
eventdir_list=sorted(glob("/home/msseo/works/72.2025_Imsil_swarm/waveforms_100km/20*"))
for eventdir in tqdm(eventdir_list):
    yyyymmddhhmmss=eventdir.split("/")[-1]
    sacfile_list=glob(eventdir+"/*SAC")
    for sacfile in sacfile_list:
        network=sacfile.split("/")[-1].split(".")[0]
        station=sacfile.split("/")[-1].split(".")[1]
        channel=sacfile.split("/")[-1].split(".")[3]
        new_sacfile=f"/home/msseo/works/72.2025_Imsil_swarm/waveforms_100km/{yyyymmddhhmmss}/{yyyymmddhhmmss}.{network}.{station}.{channel}.sac"
        shutil.copyfile(sacfile, new_sacfile)
```
