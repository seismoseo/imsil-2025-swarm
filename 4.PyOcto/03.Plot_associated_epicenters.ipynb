## Import libraries

```python
# Import libraries for matrix and vector computation
import numpy as np
from numpy.linalg import norm
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import pygmt as pmt # for plotting
```

## Load `PyOcto` event catalog

```python
event_table_251102=pd.read_csv("/home/msseo/works/72.2025_Imsil_swarm/4.PyOcto/251102/pyocto_kim1983_prob0.6.csv", sep=",")
event_table_251103=pd.read_csv("/home/msseo/works/72.2025_Imsil_swarm/4.PyOcto/251103/pyocto_kim1983_prob0.6.csv", sep=",")
event_table_combined=pd.concat([event_table_251102, event_table_251103]).reset_index(drop=True)
event_table_combined
```

## Plotting

```python
# Known hypocenter of Imsil M 1.9 event
evla, evlo=35.5498, 127.2947
```

```python
fig=pmt.Figure()
pmt.config(FORMAT_GEO_MAP="ddd.x", MAP_FRAME_TYPE="plain") # Use decimal degrees for latitude/longitude

region=[126.5, 128.0, 34.8, 36.0]

fig.basemap(region=region, projection="M15c", frame=["a", '+t"PyOcto locations (Prob. > 0.6)"'])
fig.coast(land="white", water="skyblue" ,shorelines=True, map_scale="n0.77/0.87+l+c127.29/35.55+w20")

# Plot epicenters
csize=0.2*np.exp(0.5*1.5)
fig.plot(x=event_table_combined["longitude"], y=event_table_combined["latitude"], style=f"c{csize}c", pen="1.5p,red")
fig.plot(x=evlo, y=evla, style="x0.5c", pen="2p,blue")
    
fig.show(width=500)
# fig.savefig("./Figures/Kim2011_all_events.pdf", dpi=500)
```

```python
fig=pmt.Figure()
pmt.config(FORMAT_GEO_MAP="ddd.xx", MAP_FRAME_TYPE="plain") # Use decimal degrees for latitude/longitude

region=[127.1, 127.5, 35.4, 35.7]

fig.basemap(region=region, projection="M15c", frame=["a", '+t"PyOcto locations (Prob. > 0.6)"'])
fig.coast(land="white", water="skyblue" ,shorelines=True, map_scale="n0.5/0.87+l+c127.29/35.55+w5")

# Plot epicenters
csize=0.2*np.exp(0.5*1.5)
fig.plot(x=event_table_combined["longitude"], y=event_table_combined["latitude"], style=f"c{csize}c", pen="1.5p,red")
fig.plot(x=evlo, y=evla, style="x0.5c", pen="2p,blue")
    
fig.show(width=500)
# fig.savefig("./Figures/Kim2011_all_events.pdf", dpi=500)
```
