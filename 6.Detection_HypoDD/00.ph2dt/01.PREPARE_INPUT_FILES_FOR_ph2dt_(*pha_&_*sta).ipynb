## Read `*.arc` file and find the lines containing `ES` at $47^{\mathrm{th}}$ column

```python
import numpy as np
```

```python
f1=open("Imsil.arc", 'r')
lines=f1.readlines()
f2=open("Imsil_dd.arc", 'w')
for line in lines:
    if line[46:48]=="ES":
        f2.write(line[:11]+"N"+line[12:])
    else:
        f2.write(line)
f2.close()
f1.close()
```

**Afterwards, use `ncsn2pha` to convert `*_dd.arc` to `*.pha` file**

## Convert `*.sta` file to `station.dat`

```python
f1=open("Imsil_hyp.sta", 'r')
f2=open("station.dat", 'w')
lines=f1.readlines()
for line in lines:
    network, station=line[6:8], line[0:5].replace(" ", "")
    lat_deg, lat_min=int(line[15:17]), float(line[18:25])
    lon_deg, lon_min=int(line[26:29]), float(line[30:37])
    latitude, longitude=np.round(lat_deg+lat_min/60, 5), np.round(lon_deg+lon_min/60, 5)
    f2.write(network+station+" "+str(latitude)+" "+str(longitude)+"\n")
f2.close()
f1.close()
```
